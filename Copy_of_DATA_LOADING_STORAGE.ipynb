```python
# Demonstrating skiprows, nrows, usecols, index_col, and names parameters in pd.read_csv

# 1. usecols:
print("usecols")
df_usecols = pd.read_csv('employees.csv', usecols=['Employee_ID', 'Name', 'Salary'])
display(df_usecols)

# 2. nrows
print("nrows")
df_nrows = pd.read_csv('employees.csv', nrows=3)
display(df_nrows)

# 3. skiprows:
print("skiprows")
df_skiprows = pd.read_csv('employees.csv', skiprows=[1, 2])
display(df_skiprows)

# 4. index_col:
print("index_col")
df_index = pd.read_csv('employees.csv', index_col='Employee_ID')
display(df_index)

# 5. names:
print("names")
custom_names = ['ID', 'Emp_Name', 'Dept', 'Location', 'Annual_Pay']
df_names = pd.read_csv('employees.csv', skiprows=1, names=custom_names)
display(df_names)
```

```python
cursor=conn.cursor()
print("Database created Successfully")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS student_table(

Name TEXT,
ROLL_NO INTEGER,
Marks INTEGER)



""")
print("Table is created successfully")
```

```python
# write the data into table
import sqlite3
conn=sqlite3.connect("students.db")
cursor=conn.cursor()
cursor.execute(
    "INSERT INTO student_table VALUES(?,?,?)",
    ("Asmit",101,90)
)
cursor.execute(
    "INSERT INTO student_table VALUES(?,?,?)",
    ("Abhiyank",102,80)
)
cursor.execute(
    "INSERT INTO student_table VALUES(?,?,?)",
    ("Ayansh",103,70)
)

```

```python
conn.commit()
print("Record inserted successfully ")
conn.close()
```

```python
# Retrive|read the data
import sqlite3
conn=sqlite3.connect("students.db")
cursor=conn.cursor()
cursor.execute("SELECT * FROM student_table")
result=cursor.fetchall()
for row in result:
    print(row)
conn.close()
```
